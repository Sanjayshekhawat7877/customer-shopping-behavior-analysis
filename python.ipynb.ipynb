```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/amlanmohanty1/customer-trends-data-analysis-SQL-Python-PowerBI/refs/heads/main/customer_shopping_behavior.csv")

print(df.head())  # Check first 5 rows

```

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df['review rating'] = ( df.groupby('category')['review rating'] .transform(lambda x: x.fillna(x.median())))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a columns age_group
labels=['Young Adult','Adult','Middle-aged','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days
# Step 1: lowercase column names and replace spaces
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# Step 2: define mapping with lowercase keys
frequency_mapping = {
    'fortnightly': 14,
    'weekly': 7,
    'monthly': 30,
    'quarterly': 90,
    'bi-weekly': 14,
    'annually': 365,
    'every 3 months': 90
}

# Step 3: create new column
df['purchase_frequency_days'] = df['frequency_of_purchases'].str.lower().map(frequency_mapping)

# Step 4: check
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```
